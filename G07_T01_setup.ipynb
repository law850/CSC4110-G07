### 1d) Understand the problem
The models will be used by **Mvua Alert** (a weather-alert start-up in Kisumu) to send SMS warnings to farmers and fishers. The first model will estimate the day's maximum temperature from 9 a.m. readings so farmers can plan their activities, and the second model will predict if it will rain tomorrow so fishers can avoid going out to the lake in dangerous weather.

* **Regression Target:** `MaxTemp` (Maximum temperature in °C). Best predictors: `Temp9am`, `Humidity9am`, `MinTemp`.
* **Classification Target:** `RainTomorrow` (Yes/No). Best predictors: `RainToday`, `Humidity3pm`, `Pressure3pm`, `WindGustSpeed`. **bold text**
